```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    InputLayer, Reshape, Conv2D,
    MaxPooling2D, Flatten, Dense
)

# --------------------------------------------------
# 1. DATASET
# --------------------------------------------------
mnist = tf.keras.datasets.mnist
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# normalize (0–1)
train_images = train_images / 255.0
test_images  = test_images / 255.0

print("Train shape:", train_images.shape)
print("Test shape :", test_images.shape)

# --------------------------------------------------
# 2. MODEL (CNN)
# --------------------------------------------------
model = Sequential()

model.add(InputLayer(input_shape=(28, 28)))
model.add(Reshape((28, 28, 1)))

model.add(Conv2D(32, kernel_size=(3,3), activation="relu"))
model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Conv2D(64, kernel_size=(3,3), activation="relu"))
model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Flatten())
model.add(Dense(128, activation="relu"))
model.add(Dense(10))  # logits (0–9)

model.compile(
    optimizer="adam",
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=["accuracy"]
)

model.summary()

# --------------------------------------------------
# 3. TRAIN
# --------------------------------------------------
history = model.fit(
    train_images,
    train_labels,
    epochs=10,
    batch_size=64,
    validation_split=0.1,
    verbose=1
)

# --------------------------------------------------
# 4. EVALUATE
# --------------------------------------------------
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=0)
print("Test accuracy:", test_acc)

# --------------------------------------------------
# 5. SINGLE IMAGE PREDICTION
# --------------------------------------------------
```

```python
index = 999  # istediğin index
img = train_images[index]          # (28, 28)
img_batch = img.reshape(1, 28, 28) # batch ekle

prediction = model.predict(img_batch)
predicted_class = np.argmax(prediction, axis=1)[0]

print("Predicted:", predicted_class)
print("True label:", train_labels[index])

# --------------------------------------------------
# 6. VISUALIZE
# --------------------------------------------------
plt.imshow(img, cmap="gray")
plt.title(f"Predicted: {predicted_class}")
plt.axis("off")
plt.show()
```
