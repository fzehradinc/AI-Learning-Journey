# Imputation - bos verileri doldurmanin yollari, orta deger, en yakin komsulari, resgression

```python
from sklearn.impute import SimpleImputer
```

```python
imp=SimpleImputer(strategy='mean')
```

```python
yenidata=imp.fit_transform(df)
```

```python
pd.DataFrame(yenidata)
```

```python
from sklearn.impute import KNNImputer
```

```python
knn=KNNImputer()
```

```python
dolu=knn.fit_transform(df)
```

```python
pd.DataFrame(dolu)
```

```python
from sklearn.experimental import enable_iterative_imputer
```

```python
from sklearn.impute import IterativeImputer
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
imp=IterativeImputer(estimator=lr)
```

```python
lrdata=imp.fit_transform(df)
```

```python
pd.DataFrame(lrdata)
```
