```python
import wikipediaapi

# Wikipedia API'sini başlat ve kullanıcı aracısını belirtin
wiki_wiki = wikipediaapi.Wikipedia(
    user_agent="YourCustomUserAgent",
    language='tr'
)

# Amerika Birleşik Devletleri sayfasını çek
page_py = wiki_wiki.page("Amerika_Birleşik_Devletleri")

# Sayfanın içeriğini yazdır
print(page_py.title)
f = print(page_py.summary[:500])  # İlk 500 karakter
```

```python
text = f.read()
doc = nlp(text)
sentence1 = list(doc.sents)[0]
print (sentence1)
```
