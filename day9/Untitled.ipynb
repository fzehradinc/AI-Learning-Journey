```python
page=requests.get('https://www.karar.com/yazarlar/zafer-acar')
```

```python
soup = BeautifulSoup(page.content, "html.parser")
```

```python
articles = soup.select("article.item.box-shadow")
```

```python
articles = soup.find_all("article", class_=["item", "box-shadow"])


```

```python
links = []

for art in articles:
    a = art.find("a")
    if a and a.get("href"):
        links.append("https://www.karar.com" + a["href"])

links[:30]
```
