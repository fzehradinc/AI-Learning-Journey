```python
df=pd.read_csv('yelp.csv')
```

```python
df.head()
```

```python
df['text']
```

```python
df.shape
```

```python
df['text']=df['text'].str.lower()
```

```python
df['text'][0]
```

```python
# küçük harfe çevir
df['text'] = df['text'].str.lower()

# noktalama işaretlerini kaldır
df['text'] = df['text'].str.replace(r'[^\w\s]', '', regex=True)

# rakamları kaldır
df['text'] = df['text'].str.replace(r'\d+', '', regex=True)

# satır sonlarını (\n) kaldır
df['text'] = df['text'].str.replace('\n', '', regex=True)

# enter / carriage return (\r) kaldır
df['text'] = df['text'].str.replace('\r', '', regex=True)
```

```python
df['text'][0]
```

```python
## Sentimental Analysis
```

```python
import seaborn as sns
```

```python
sns.countplot(x=df['stars']);
```

```python
ybw=df[(df.stars==1) | (df.stars==5)]
```

```python
ybw.reset_index(drop=True,inplace=True)
```

```python
ybw.head()
```

```python
x=ybw[['text']]
y=ybw[['stars']]
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.metrics import accuracy_score
```

```python
def ekkok(text):
    words=TextBlob(text).words
    return [word.lemmatize() for word in words]
```

```python
vect=CountVectorizer(analyzer=ekkok, stop_words='english', ngram_range=(1,3))
```

```python
yenix=vect.fit_transform(x['text'])
```

```python
tf=pd.DataFrame(vect.fit_transform(x['text']).toarray(), columns=vect.get_feature_names_out())
```

```python
tf.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(yenix,y,random_state=42, test_size=.10)
```

```python
dt=DecisionTreeClassifier()
```

```python
model=dt.fit(x_train, y_train)
```

```python
tahmin=model.predict(x_test)
```

```python
accuracy_score(y_test,tahmin)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
```

```python
g=GradientBoostingClassifier()
```

```python
g.fit(x_train,y_train)
```

```python
gtahmin=g.predict(x_test)
```

```python
accuracy_score(y_test,gtahmin)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
model=lr.fit(x_train, y_train)
```
