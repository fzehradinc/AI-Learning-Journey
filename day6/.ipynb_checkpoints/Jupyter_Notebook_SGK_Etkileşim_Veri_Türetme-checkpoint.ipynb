````python
# ============================================================================
# BLOK 1: ORTAM HAZIRLIĞI + VERİ YÜKLEME + İLK İNCELEME
# ============================================================================

import pandas as pd
import json
import re
from collections import defaultdict
import warnings
warnings.filterwarnings('ignore')

# Pandas ayarları
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 50)
pd.set_option('display.width', 1000)

print("="*80)
print("BLOK 1: VERİ YÜKLEME VE İLK İNCELEME")
print("="*80)

# ============================================================================
# AŞAMA 1.1: SGK JSON Dosyasını Yükle
# ============================================================================
print("\n📂 SGK JSON dosyası yükleniyor...")
with open('sgk_drugs_unique.json', 'r', encoding='utf-8') as f:
    sgk_data = json.load(f)

sgk_df = pd.DataFrame(sgk_data)
print(f"✅ SGK'da {len(sgk_df):,} ürün yüklendi")

# ============================================================================
# AŞAMA 1.2: DDI CSV Dosyasını Yükle
# ============================================================================
print("\n📂 DDI CSV dosyası yükleniyor...")
ddi_df = pd.read_csv('DDI_data.csv')
print(f"✅ DDI'da {len(ddi_df):,} etkileşim yüklendi")

# ============================================================================
# AŞAMA 1.3: SGK Verisini İncele
# ============================================================================
print("\n" + "="*80)
print("SGK VERİSİ İNCELEMESİ")
print("="*80)

print(f"\n📊 Temel Bilgiler:")
print(f"  Toplam satır: {len(sgk_df):,}")
print(f"  Sütun sayısı: {len(sgk_df.columns)}")

print(f"\n📋 Sütunlar:")
for col in sgk_df.columns:
    print(f"  - {col}")

print(f"\n❓ Kritik Sütunlardaki Eksik Değerler:")
critical_cols = ['drug_name', 'active_ingredient', 'skrs_atc_name', 'tick_atc_code']
for col in critical_cols:
    if col in sgk_df.columns:
        null_count = sgk_df[col].isnull().sum()
        null_pct = (null_count / len(sgk_df)) * 100
        print(f"  {col:<25} {null_count:>8,} ({null_pct:>5.1f}%)")

# KRİTİK: Çoklu etken madde kontrolü
print(f"\n⚠️ ÇOKLU ETKEN MADDE KONTROLÜ:")
if 'active_ingredient' in sgk_df.columns:
    has_comma = sgk_df['active_ingredient'].astype(str).str.contains(',', na=False)
    combo_count = has_comma.sum()
    combo_pct = (combo_count / len(sgk_df)) * 100
    print(f"  Virgüllü (kombine) ürün sayısı: {combo_count:,} ({combo_pct:.1f}%)")
    
    if combo_count > 0:
        print(f"\n  📝 Örnek kombine ürünler:")
        combo_examples = sgk_df[has_comma][['drug_name', 'active_ingredient']].head(3)
        for idx, row in combo_examples.iterrows():
            print(f"    • {row['drug_name'][:60]}...")
            print(f"      Etken maddeler: {row['active_ingredient']}")

# skrs_atc_name incelemesi
print(f"\n🔍 SKRS_ATC_NAME (İngilizce etken madde) Durumu:")
if 'skrs_atc_name' in sgk_df.columns:
    non_null = sgk_df['skrs_atc_name'].notna().sum()
    non_null_pct = (non_null / len(sgk_df)) * 100
    print(f"  Dolu olan: {non_null:,} ({non_null_pct:.1f}%)")
    
    if non_null > 0:
        print(f"\n  📝 Örnek skrs_atc_name değerleri:")
        samples = sgk_df['skrs_atc_name'].dropna().head(5).tolist()
        for i, sample in enumerate(samples, 1):
            print(f"    {i}. {sample}")

# ============================================================================
# AŞAMA 1.4: DDI Verisini İncele
# ============================================================================
print("\n" + "="*80)
print("DDI VERİSİ İNCELEMESİ")
print("="*80)

print(f"\n📊 Temel Bilgiler:")
print(f"  Toplam etkileşim: {len(ddi_df):,}")
print(f"  Sütunlar: {list(ddi_df.columns)}")

print(f"\n🔢 Benzersiz İlaç Sayıları:")
print(f"  drug1_name: {ddi_df['drug1_name'].nunique():,}")
print(f"  drug2_name: {ddi_df['drug2_name'].nunique():,}")

print(f"\n📝 Örnek DDI Kayıtları:")
print(ddi_df[['drug1_name', 'drug2_name', 'interaction_type']].head(3))

print(f"\n🔝 En Sık 5 Etkileşim Tipi:")
top_interactions = ddi_df['interaction_type'].value_counts().head(5)
for itype, count in top_interactions.items():
    pct = (count / len(ddi_df)) * 100
    print(f"  {itype[:50]:<50} {count:>8,} ({pct:>5.1f}%)")

# ============================================================================
# BLOK 1 ÖZET
# ============================================================================
print("\n" + "="*80)
print("✅ BLOK 1 TAMAMLANDI - VERİLER YÜKLENDI VE İNCELENDİ")
print("="*80)
print(f"\n📊 Özet:")
print(f"  SGK Ürün Sayısı: {len(sgk_df):,}")
print(f"  DDI Etkileşim Sayısı: {len(ddi_df):,}")
if 'active_ingredient' in sgk_df.columns:
    print(f"  Kombine Ürün Oranı: {combo_pct:.1f}%")
if 'skrs_atc_name' in sgk_df.columns:
    print(f"  İngilizce Etken Madde Oranı: {non_null_pct:.1f}%")

print("\n💡 Sonraki Blok: Normalizasyon ve Etken Madde Ayrıştırma")
````

````python
# ============================================================================
# BLOK 1.5: TÜRKÇE-İNGİLİZCE ETKEN MADDE HARITALAMASI (MAPPING)
# ============================================================================

print("\n" + "="*80)
print("BLOK 1.5: TÜRKÇE-İNGİLİZCE ETKEN MADDE HARITALAMASI")
print("="*80)

# ============================================================================
# AŞAMA 1.5.1: Mevcut Durumu Analiz Et
# ============================================================================
print("\n📊 Mevcut durum analizi:")

# İngilizce etken madde bilgisi var mı?
has_skrs = sgk_df['skrs_atc_name'].notna().sum()
has_active = sgk_df['active_ingredient'].notna().sum()
missing_both = ((sgk_df['skrs_atc_name'].isna()) & (sgk_df['active_ingredient'].isna())).sum()
needs_mapping = ((sgk_df['skrs_atc_name'].isna()) & (sgk_df['active_ingredient'].notna())).sum()

print(f"  skrs_atc_name dolu (İngilizce):  {has_skrs:>8,} ({has_skrs/len(sgk_df)*100:>5.1f}%)")
print(f"  active_ingredient dolu (Türkçe): {has_active:>8,} ({has_active/len(sgk_df)*100:>5.1f}%)")
print(f"  Her ikisi de boş:                {missing_both:>8,} ({missing_both/len(sgk_df)*100:>5.1f}%)")
print(f"  MAPPING GEREKENler:              {needs_mapping:>8,} ({needs_mapping/len(sgk_df)*100:>5.1f}%)")

# ============================================================================
# AŞAMA 1.5.2: Türkçe-İngilizce Çeviri Sözlüğü
# ============================================================================
print("\n📚 Türkçe-İngilizce etken madde sözlüğü oluşturuluyor...")

# KAPSAMLI TÜRKÇE-İNGİLİZCE SÖZLÜK
tr_to_en_mapping = {
    # En yaygın etken maddeler
    'parasetamol': 'paracetamol',
    'asetilsalisilik asit': 'acetylsalicylic acid',
    'ibuprofen': 'ibuprofen',
    'naproksen': 'naproxen',
    'diklofenak': 'diclofenac',
    'ketoprofen': 'ketoprofen',
    'indometazin': 'indomethacin',
    
    # Antibiyotikler
    'amoksisilin': 'amoxicillin',
    'ampisilin': 'ampicillin',
    'penisilin': 'penicillin',
    'eritromisin': 'erythromycin',
    'azitromisin': 'azithromycin',
    'klaritromisin': 'clarithromycin',
    'seftriakson': 'ceftriaxone',
    'sefotaksim': 'cefotaxime',
    'sefuroksim': 'cefuroxime',
    'siprofloksasin': 'ciprofloxacin',
    'levofloksasin': 'levofloxacin',
    'moksifoksasin': 'moxifloxacin',
    'tetrasiklin': 'tetracycline',
    'doksisiklin': 'doxycycline',
    'vankomisin': 'vancomycin',
    'metronidazol': 'metronidazole',
    'trimetoprim': 'trimethoprim',
    'sulfametoksazol': 'sulfamethoxazole',
    
    # Kardiyovasküler
    'atorvastatin': 'atorvastatin',
    'simvastatin': 'simvastatin',
    'rosuvastatin': 'rosuvastatin',
    'pravastatin': 'pravastatin',
    'amlodipin': 'amlodipine',
    'nifedipin': 'nifedipine',
    'diltiazem': 'diltiazem',
    'verapamil': 'verapamil',
    'enalapril': 'enalapril',
    'ramipril': 'ramipril',
    'lisinopril': 'lisinopril',
    'losartan': 'losartan',
    'valsartan': 'valsartan',
    'irbesartan': 'irbesartan',
    'metoprolol': 'metoprolol',
    'bisoprolol': 'bisoprolol',
    'atenolol': 'atenolol',
    'karvedilol': 'carvedilol',
    'furosemid': 'furosemide',
    'hidroklorotiyazid': 'hydrochlorothiazide',
    'spironolakton': 'spironolactone',
    'warfarin': 'warfarin',
    'heparin': 'heparin',
    'enoksaparin': 'enoxaparin',
    'aspirin': 'aspirin',
    'klopidogrel': 'clopidogrel',
    
    # Diyabet
    'metformin': 'metformin',
    'gliklazid': 'gliclazide',
    'glimepirid': 'glimepiride',
    'insulin': 'insulin',
    'insülin': 'insulin',
    'sitagliptin': 'sitagliptin',
    'empagliflozin': 'empagliflozin',
    
    # GI sistem
    'omeprazol': 'omeprazole',
    'lansoprazol': 'lansoprazole',
    'pantoprazol': 'pantoprazole',
    'rabeprazol': 'rabeprazole',
    'esomeprazol': 'esomeprazole',
    'ranitidin': 'ranitidine',
    'famotidin': 'famotidine',
    'metoklopramid': 'metoclopramide',
    'domperidon': 'domperidone',
    
    # Solunum sistemi
    'salbutamol': 'salbutamol',
    'formoterol': 'formoterol',
    'salmeterol': 'salmeterol',
    'budesonid': 'budesonide',
    'flutikazon': 'fluticasone',
    'beklometazon': 'beclomethasone',
    'montelukast': 'montelukast',
    'teofilin': 'theophylline',
    
    # Santral sinir sistemi
    'diazepam': 'diazepam',
    'alprazolam': 'alprazolam',
    'lorazepam': 'lorazepam',
    'klonazepam': 'clonazepam',
    'fluoksetin': 'fluoxetine',
    'sertralin': 'sertraline',
    'paroksetin': 'paroxetine',
    'essitalopram': 'escitalopram',
    'venlafaksin': 'venlafaxine',
    'amitriptilin': 'amitriptyline',
    'karbamazepin': 'carbamazepine',
    'valproik asit': 'valproic acid',
    'fenitoin': 'phenytoin',
    'levetirasetam': 'levetiracetam',
    'gabapentin': 'gabapentin',
    'pregabalin': 'pregabalin',
    'morfin': 'morphine',
    'kodein': 'codeine',
    'tramadol': 'tramadol',
    'fentanil': 'fentanyl',
    
    # Hormonlar
    'levotiroksin': 'levothyroxine',
    'propiltiyourasil': 'propylthiouracil',
    'metotreksat': 'methotrexate',
    'hidrokortizon': 'hydrocortisone',
    'prednizolon': 'prednisolone',
    'metilprednizolon': 'methylprednisolone',
    'deksametazon': 'dexamethasone',
    
    # Vitaminler ve mineraller
    'vitamin b12': 'vitamin b12',
    'vitamin d': 'vitamin d',
    'vitamin c': 'vitamin c',
    'vitamin e': 'vitamin e',
    'vitamin k': 'vitamin k',
    'folik asit': 'folic acid',
    'demir': 'iron',
    'kalsiyum': 'calcium',
    'magnezyum': 'magnesium',
    'potasyum': 'potassium',
    'çinko': 'zinc',
    
    # Diğer
    'sodyum klorür': 'sodium chloride',
    'potasyum klorür': 'potassium chloride',
    'glukoz': 'glucose',
    'mannitol': 'mannitol',
    'allopurinol': 'allopurinol',
    'kolşisin': 'colchicine',
    'asetilsistein': 'acetylcysteine',
}

print(f"✅ Sözlük hazır: {len(tr_to_en_mapping):,} benzersiz eşleştirme")

# ============================================================================
# AŞAMA 1.5.3: Mapping Fonksiyonu
# ============================================================================
print("\n🔧 Mapping fonksiyonu hazırlanıyor...")

def map_turkish_to_english(turkish_text):
    """
    Türkçe etken madde → İngilizce
    
    Args:
        turkish_text: Türkçe etken madde ismi
    
    Returns:
        str: İngilizce karşılık veya None
    """
    if pd.isna(turkish_text) or str(turkish_text).strip() == '':
        return None
    
    # Küçük harfe çevir ve temizle
    text = str(turkish_text).lower().strip()
    
    # Türkçe karakterleri çevir
    turkish_map = {
        'ç': 'c', 'ğ': 'g', 'ı': 'i', 'İ': 'i', 'i̇': 'i',
        'ö': 'o', 'ş': 's', 'ü': 'u'
    }
    for tr_char, en_char in turkish_map.items():
        text = text.replace(tr_char, en_char)
    
    # Sözlükte var mı kontrol et
    if text in tr_to_en_mapping:
        return tr_to_en_mapping[text]
    
    # Bulunamadı
    return None

# Test et
print("\n🧪 Mapping testi:")
test_cases = [
    'parasetamol',
    'ASETİLSALİSİLİK ASİT',
    'sodyum klorür',
    'bilinmeyen ilaç xyz'
]
for test in test_cases:
    result = map_turkish_to_english(test)
    print(f"  '{test}' → '{result}'")

# ============================================================================
# AŞAMA 1.5.4: SGK Verisine Mapping Uygula
# ============================================================================
print("\n" + "="*80)
print("MAPPING UYGULAMASI")
print("="*80)

print("\n📝 Türkçe etken maddeleri İngilizce'ye dönüştürüyorum...")

# Yeni sütun: ingredient_en_mapped
sgk_df['ingredient_en_mapped'] = sgk_df['active_ingredient'].apply(map_turkish_to_english)

# İstatistikler
mapped_count = sgk_df['ingredient_en_mapped'].notna().sum()
unmapped_count = ((sgk_df['ingredient_en_mapped'].isna()) & (sgk_df['active_ingredient'].notna())).sum()

print(f"\n✅ Mapping tamamlandı!")
print(f"  Başarılı mapping: {mapped_count:,} ({mapped_count/len(sgk_df)*100:.1f}%)")
print(f"  Başarısız (bilinmeyen): {unmapped_count:,} ({unmapped_count/len(sgk_df)*100:.1f}%)")

# ============================================================================
# AŞAMA 1.5.5: İngilizce Sütunu Birleştir (ÖNCELIK SİSTEMİ)
# ============================================================================
print("\n" + "="*80)
print("İNGİLİZCE SÜTUNU BİRLEŞTİRME (PRIORİTY MERGE)")
print("="*80)

print("\n🔧 Öncelik sırası ile İngilizce sütunu oluşturuluyor:")
print("   1. skrs_atc_name (orijinal İngilizce - EN GÜVENILIR)")
print("   2. ingredient_en_mapped (Türkçe'den mapping)")
print("   3. active_ingredient (Türkçe - SON ÇARE)")

# Yeni birleşik sütun oluştur
sgk_df['ingredient_en_final'] = None

# Öncelik 1: skrs_atc_name (orijinal İngilizce)
mask1 = sgk_df['skrs_atc_name'].notna()
sgk_df.loc[mask1, 'ingredient_en_final'] = sgk_df.loc[mask1, 'skrs_atc_name']

# Öncelik 2: ingredient_en_mapped (mapping'den)
mask2 = (sgk_df['ingredient_en_final'].isna()) & (sgk_df['ingredient_en_mapped'].notna())
sgk_df.loc[mask2, 'ingredient_en_final'] = sgk_df.loc[mask2, 'ingredient_en_mapped']

# Öncelik 3: active_ingredient (Türkçe - son çare)
mask3 = (sgk_df['ingredient_en_final'].isna()) & (sgk_df['active_ingredient'].notna())
sgk_df.loc[mask3, 'ingredient_en_final'] = sgk_df.loc[mask3, 'active_ingredient']

# İstatistikler
final_stats = {
    'from_skrs': (sgk_df['ingredient_en_final'] == sgk_df['skrs_atc_name']).sum(),
    'from_mapping': ((sgk_df['ingredient_en_final'] == sgk_df['ingredient_en_mapped']) & 
                     (sgk_df['ingredient_en_mapped'].notna())).sum(),
    'from_turkish': ((sgk_df['ingredient_en_final'] == sgk_df['active_ingredient']) & 
                     (sgk_df['active_ingredient'].notna()) &
                     (sgk_df['skrs_atc_name'].isna()) &
                     (sgk_df['ingredient_en_mapped'].isna())).sum(),
    'still_missing': sgk_df['ingredient_en_final'].isna().sum()
}

print(f"\n📊 Final İngilizce Sütun Kaynakları:")
print(f"  Orijinal İngilizce (skrs_atc_name): {final_stats['from_skrs']:>8,} ({final_stats['from_skrs']/len(sgk_df)*100:>5.1f}%)")
print(f"  Mapping (TR→EN sözlük):             {final_stats['from_mapping']:>8,} ({final_stats['from_mapping']/len(sgk_df)*100:>5.1f}%)")
print(f"  Türkçe (son çare):                  {final_stats['from_turkish']:>8,} ({final_stats['from_turkish']/len(sgk_df)*100:>5.1f}%)")
print(f"  Hala eksik:                         {final_stats['still_missing']:>8,} ({final_stats['still_missing']/len(sgk_df)*100:>5.1f}%)")

# ============================================================================
# AŞAMA 1.5.6: Bilinmeyen Türkçe İlaçları Tespit Et
# ============================================================================
print("\n" + "="*80)
print("BİLİNMEYEN TÜRKÇE İLAÇLARI TESPİT ET")
print("="*80)

# Bilinmeyen Türkçe ilaçları listele
unknown_turkish = sgk_df[
    (sgk_df['skrs_atc_name'].isna()) &
    (sgk_df['ingredient_en_mapped'].isna()) &
    (sgk_df['active_ingredient'].notna())
]['active_ingredient'].value_counts().head(20)

if len(unknown_turkish) > 0:
    print(f"\n⚠️ En sık karşılaşılan bilinmeyen Türkçe etken maddeler (Top 20):")
    print(f"   (Bunlar için manuel mapping eklenebilir)\n")
    for ingredient, count in unknown_turkish.items():
        print(f"  '{ingredient}': {count:>4,} ürün")
    
    print(f"\n💡 İPUCU: Bu listeyi kullanarak tr_to_en_mapping sözlüğünü genişletebilirsin!")

# ============================================================================
# AŞAMA 1.5.7: Başarılı Mapping Örnekleri
# ============================================================================
print("\n📝 Örnek Başarılı Mappingler:")
successful_mappings = sgk_df[sgk_df['ingredient_en_mapped'].notna()].head(10)
for idx, row in successful_mappings.iterrows():
    print(f"  '{row['active_ingredient'][:40]}' → '{row['ingredient_en_mapped'][:40]}'")

# ============================================================================
# BLOK 1.5 ÖZET
# ============================================================================
print("\n" + "="*80)
print("✅ BLOK 1.5 TAMAMLANDI - MAPPING BİTTİ")
print("="*80)

total_english = final_stats['from_skrs'] + final_stats['from_mapping']
total_english_pct = total_english / len(sgk_df) * 100
original_english_pct = has_skrs / len(sgk_df) * 100

print(f"\n🎯 İYİLEŞTİRME:")
print(f"  Önceki İngilizce oran: {original_english_pct:.1f}%")
print(f"  Sonraki İngilizce oran: {total_english_pct:.1f}%")
print(f"  Artış: +{total_english_pct - original_english_pct:.1f}%")

print(f"\n📊 Yeni ingredient_en_final sütunu kullanılacak!")
print(f"   Bu sütun BLOK 2'de normalizasyon için kullanılacak.")

print("\n💡 Sonraki Blok: BLOK 2 - Normalizasyon (ingredient_en_final kullanacak)")
````

````python
# ============================================================================
# BLOK 2: NORMALİZASYON + ETKEN MADDE AYRIŞTIRMA
# ============================================================================

print("\n" + "="*80)
print("BLOK 2: NORMALİZASYON VE ETKEN MADDE AYRIŞTIRMA")
print("="*80)

# ============================================================================
# AŞAMA 2.1: Normalizasyon Fonksiyonu (DEĞİŞMEDİ)
# ============================================================================
print("\n📝 Normalizasyon fonksiyonu hazırlanıyor...")

def normalize_ingredient(text):
    if pd.isna(text) or text == '' or str(text).lower() == 'nan':
        return None
    
    normalized = str(text).lower().strip()
    
    # TÜRKÇE KARAKTERLERİ ÇEVİR
    turkish_map = {
        'ç': 'c', 'ğ': 'g', 'ı': 'i', 'İ': 'i', 'i̇': 'i',
        'ö': 'o', 'ş': 's', 'ü': 'u'
    }
    for tr_char, en_char in turkish_map.items():
        normalized = normalized.replace(tr_char, en_char)
    
    # Özel karakterleri kaldır
    normalized = re.sub(r'[^a-z0-9\s-]', '', normalized)
    normalized = re.sub(r'\s+', ' ', normalized)
    normalized = normalized.strip()
    
    return normalized if normalized else None

# Fonksiyonu test et
print("✅ Normalizasyon fonksiyonu hazır")

# ============================================================================
# AŞAMA 2.2: DDI Verisini Normalize Et (DEĞİŞMEDİ)
# ============================================================================
print("\n" + "="*80)
print("DDI NORMALİZASYONU")
print("="*80)

print("\n📝 DDI etken maddelerini normalize ediyorum...")

# drug1_name ve drug2_name'i normalize et
ddi_df['ingredient_a_norm'] = ddi_df['drug1_name'].apply(normalize_ingredient)
ddi_df['ingredient_b_norm'] = ddi_df['drug2_name'].apply(normalize_ingredient)

# Null olanları temizle
before_clean = len(ddi_df)
ddi_df = ddi_df[
    ddi_df['ingredient_a_norm'].notna() & 
    ddi_df['ingredient_b_norm'].notna()
].copy()
after_clean = len(ddi_df)

print(f"✅ DDI normalize edildi")
print(f"  Önceki satır sayısı: {before_clean:,}")
print(f"  Sonraki satır sayısı: {after_clean:,}")
print(f"  Temizlenen: {before_clean - after_clean:,}")

# Benzersiz etken madde sayısı
all_ddi_ingredients = set(pd.concat([
    ddi_df['ingredient_a_norm'],
    ddi_df['ingredient_b_norm']
]).unique())

print(f"\n🔢 DDI'da benzersiz normalize etken madde sayısı: {len(all_ddi_ingredients):,}")

# ============================================================================
# AŞAMA 2.3: SGK Verisini Normalize Et
# 🔴 BU KISIM GÜNCELLENDİ - ingredient_en_final KULLANILACAK
# ============================================================================
print("\n" + "="*80)
print("SGK NORMALİZASYONU")
print("="*80)

print("\n📝 SGK etken maddelerini normalize ediyorum...")
print("   Kaynak: ingredient_en_final (Blok 1.5'ten gelen)")

# 🔴 YENİ: Basitleştirilmiş normalizasyon
sgk_df['ingredient_en_norm'] = sgk_df['ingredient_en_final'].apply(normalize_ingredient)

# İstatistikler
normalized_count = sgk_df['ingredient_en_norm'].notna().sum()
failed_count = sgk_df['ingredient_en_norm'].isna().sum()

print(f"\n✅ SGK normalize edildi!")
print(f"  Başarılı normalizasyon: {normalized_count:,} ({normalized_count/len(sgk_df)*100:.1f}%)")
print(f"  Başarısız (eksik veri): {failed_count:,} ({failed_count/len(sgk_df)*100:.1f}%)")

# ============================================================================
# AŞAMA 2.4: KOMBİNE ÜRÜNLERİ AYIR (EXPAND) (DEĞİŞMEDİ)
# ============================================================================
print("\n" + "="*80)
print("KOMBİNE ÜRÜNLERİ AYRIŞTIRMA (EXPAND)")
print("="*80)

print("\n⚠️ Bu aşama kritik! Virgüllü etken maddeleri ayırıp her biri için ayrı satır oluşturacağız.")

sgk_expanded_rows = []
combo_products = 0
single_products = 0

for idx, row in sgk_df.iterrows():
    original_ingredient = row.get('active_ingredient', '')
    
    # Virgül var mı kontrol et
    if pd.notna(original_ingredient) and ',' in str(original_ingredient):
        # ÇOKLU ETKEN MADDE VAR - Ayır!
        combo_products += 1
        parts = str(original_ingredient).split(',')
        
        for part in parts:
            part_normalized = normalize_ingredient(part.strip())
            
            if part_normalized:  # Boş değilse
                row_copy = row.to_dict()
                row_copy['ingredient_en_norm'] = part_normalized
                row_copy['is_combination'] = True
                row_copy['original_combination'] = original_ingredient
                row_copy['combination_part'] = part.strip()
                sgk_expanded_rows.append(row_copy)
    else:
        # TEK ETKEN MADDE - Olduğu gibi ekle
        single_products += 1
        row_dict = row.to_dict()
        row_dict['is_combination'] = False
        row_dict['original_combination'] = None
        row_dict['combination_part'] = None
        sgk_expanded_rows.append(row_dict)
    
    # İlerleme göstergesi
    if (idx + 1) % 5000 == 0:
        print(f"  İşlenen: {idx + 1:,} / {len(sgk_df):,}...")

# Yeni DataFrame oluştur
sgk_expanded_df = pd.DataFrame(sgk_expanded_rows)

print(f"\n✅ Kombine ürünler ayrıştırıldı!")
print(f"\n📊 Sonuçlar:")
print(f"  Orijinal SGK satır sayısı: {len(sgk_df):,}")
print(f"  Expanded SGK satır sayısı: {len(sgk_expanded_df):,}")
print(f"  Fark (ekstra satırlar): {len(sgk_expanded_df) - len(sgk_df):,}")
print(f"\n  Tek etken maddeli ürün: {single_products:,}")
print(f"  Kombine ürün (ayrıştırıldı): {combo_products:,}")
print(f"  Kombine ürünlerden gelen toplam satır: {sgk_expanded_df['is_combination'].sum():,}")

# Benzersiz etken madde sayısı
unique_ingredients_expanded = sgk_expanded_df['ingredient_en_norm'].dropna().nunique()
print(f"\n🔢 Benzersiz normalize etken madde sayısı (expanded): {unique_ingredients_expanded:,}")

# ============================================================================
# BLOK 2 ÖZET
# ============================================================================
print("\n" + "="*80)
print("✅ BLOK 2 TAMAMLANDI - NORMALİZASYON VE AYRIŞTIRMA BİTTİ")
print("="*80)

print(f"\n📊 Özet İstatistikler:")
print(f"\n  DDI Veritabanı:")
print(f"    Normalize etkileşim sayısı: {len(ddi_df):,}")
print(f"    Benzersiz etken madde: {len(all_ddi_ingredients):,}")

print(f"\n  SGK Veritabanı (Expanded):")
print(f"    Toplam satır: {len(sgk_expanded_df):,}")
print(f"    Benzersiz etken madde: {unique_ingredients_expanded:,}")
print(f"    Kombine ürünlerden gelen satır: {sgk_expanded_df['is_combination'].sum():,}")

print("\n💡 Sonraki Blok: Kapsama Analizi (Feasibility)")
```

---

## 📋 **ÖZET - YAPMAN GEREKENLER:**

### ✅ **1. BLOK 1.5'i Çalıştır (Yeni Kod)**
- Yukarıdaki **yeni Blok 1.5 kodunu** BLOK 1'den sonra çalıştır
- Hiç kütüphane yüklemeye gerek yok
- Mapping sözlüğü otomatik çalışacak

### ✅ **2. BLOK 2'yi Güncelle**
- **AŞAMA 2.3'teki** `get_ingredient_normalized` fonksiyonunu **SİL**
- Yerine basit hali koy: `sgk_df['ingredient_en_norm'] = sgk_df['ingredient_en_final'].apply(normalize_ingredient)`

### ✅ **3. Pipeline'ı Çalıştır**
- BLOK 1.5 → BLOK 2 → BLOK 3 → ... sırasıyla çalıştır

---

## 🎯 **BEKLENEN SONUÇ:**
```
ÖNCE:
❌ İngilizce: %31.2
❌ Kapsama: %17.9
❌ Türetilen DDI: 1.3M

SONRA:
✅ İngilizce: ~%70-80 (mapping + orijinal)
✅ Kapsama: ~%35-40
✅ Türetilen DDI: ~2.5-3M

2x-2.5x ARTIŞ BEKLENİYOR! 🚀
````

````python
# ============================================================================
# BLOK 3: KAPSAMA ANALİZİ (FEASIBILITY ANALYSIS)
# ============================================================================

print("\n" + "="*80)
print("BLOK 3: KAPSAMA ANALİZİ (FEASIBILITY)")
print("="*80)

# ============================================================================
# AŞAMA 3.1: Benzersiz Ingredient Setlerini Oluştur
# ============================================================================
print("\n📊 Benzersiz ingredient setlerini oluşturuyorum...")

# DDI'dan tüm benzersiz etken maddeler
ddi_ingredients_set = set(pd.concat([
    ddi_df['ingredient_a_norm'],
    ddi_df['ingredient_b_norm']
]).dropna().unique())

# SGK'dan tüm benzersiz etken maddeler (EXPANDED versiyondan)
sgk_ingredients_set = set(sgk_expanded_df['ingredient_en_norm'].dropna().unique())

print(f"✅ Setler hazır")
print(f"  DDI benzersiz ingredient: {len(ddi_ingredients_set):,}")
print(f"  SGK benzersiz ingredient: {len(sgk_ingredients_set):,}")

# ============================================================================
# AŞAMA 3.2: Örtüşme ve Farkları Hesapla
# ============================================================================
print("\n" + "="*80)
print("ÖRTÜŞME ANALİZİ")
print("="*80)

# Kesişim (overlap): Her iki sette de olan ingredient'lar
overlap = ddi_ingredients_set & sgk_ingredients_set

# Sadece SGK'da olan (DDI'da yok)
sgk_only = sgk_ingredients_set - ddi_ingredients_set

# Sadece DDI'da olan (SGK'da yok)
ddi_only = ddi_ingredients_set - sgk_ingredients_set

print(f"\n📊 Venn Diagram İstatistikleri:")
print(f"  DDI toplam ingredient:        {len(ddi_ingredients_set):>10,}")
print(f"  SGK toplam ingredient:        {len(sgk_ingredients_set):>10,}")
print(f"  ÖRTÜŞEN (overlap) ingredient: {len(overlap):>10,}")
print(f"  Sadece SGK'da (DDI'da yok):   {len(sgk_only):>10,}")
print(f"  Sadece DDI'da (SGK'da yok):   {len(ddi_only):>10,}")

# ============================================================================
# AŞAMA 3.3: Coverage Rate Hesapla
# ============================================================================
print("\n" + "="*80)
print("COVERAGE (KAPSAMA) ORANLARI")
print("="*80)

# SGK Coverage: SGK ingredient'larının yüzde kaçı DDI'da var?
sgk_coverage_pct = (len(overlap) / len(sgk_ingredients_set) * 100) if len(sgk_ingredients_set) > 0 else 0

# DDI Coverage: DDI ingredient'larının yüzde kaçı SGK'da var?
ddi_coverage_pct = (len(overlap) / len(ddi_ingredients_set) * 100) if len(ddi_ingredients_set) > 0 else 0

print(f"\n📈 Ingredient Bazında Kapsama:")
print(f"  SGK ingredient'ların DDI'da bulunma oranı: {sgk_coverage_pct:>6.2f}%")
print(f"  DDI ingredient'ların SGK'da bulunma oranı: {ddi_coverage_pct:>6.2f}%")

# ============================================================================
# AŞAMA 3.4: Ürün Bazında Kapsama (Product-Level Coverage)
# ============================================================================
print("\n" + "="*80)
print("ÜRÜN BAZINDA KAPSAMA")
print("="*80)

print(f"\n📦 Ürün düzeyinde eşleşme analizi...")

# Etkileşim türetebileceğimiz ürünler (ingredient'i overlap'te olan)
sgk_matched = sgk_expanded_df[sgk_expanded_df['ingredient_en_norm'].isin(overlap)]
sgk_unmatched = sgk_expanded_df[~sgk_expanded_df['ingredient_en_norm'].isin(overlap)]

# Ürün eşleşme oranı
product_match_count = len(sgk_matched)
product_match_pct = (product_match_count / len(sgk_expanded_df) * 100) if len(sgk_expanded_df) > 0 else 0

print(f"\n📊 Ürün Eşleşme İstatistikleri:")
print(f"  Toplam SGK ürün (satır):      {len(sgk_expanded_df):>10,}")
print(f"  DDI'da eşleşen ürün:          {product_match_count:>10,}")
print(f"  DDI'da eşleşMEyen ürün:       {len(sgk_unmatched):>10,}")
print(f"  Ürün eşleşme oranı:           {product_match_pct:>9.2f}%")

# Orijinal ürün bazında (combo expand öncesi)
sgk_original_matched = sgk_df[sgk_df['ingredient_en_norm'].isin(overlap)]
original_product_match_pct = (len(sgk_original_matched) / len(sgk_df) * 100) if len(sgk_df) > 0 else 0

print(f"\n  Orijinal ürün bazında (combo öncesi):")
print(f"    Eşleşen orijinal ürün:      {len(sgk_original_matched):>10,}")
print(f"    Orijinal ürün eşleşme oranı:{original_product_match_pct:>9.2f}%")

# ============================================================================
# AŞAMA 3.5: Tahmini Etkileşim Sayısı (Projection Estimate)
# ============================================================================
print("\n" + "="*80)
print("TAHMİNİ ETKİLEŞİM SAYISI")
print("="*80)

print(f"\n🔮 Türetilecek etkileşim tahmini...")

# Her eşleşen SGK ingredient için DDI'daki ortalama etkileşim sayısı
avg_interactions_per_ingredient = len(ddi_df) / len(ddi_ingredients_set)

# Tahmini toplam etkileşim
estimated_interactions = int(product_match_count * avg_interactions_per_ingredient)

print(f"\n📊 Tahmin:")
print(f"  Eşleşen ürün sayısı:           {product_match_count:>10,}")
print(f"  Ingredient başına ort. etkileşim: {avg_interactions_per_ingredient:>10,.1f}")
print(f"  TAHMİNİ toplam etkileşim:      {estimated_interactions:>10,}")

# ============================================================================
# AŞAMA 3.6: Kalite Değerlendirmesi
# ============================================================================
print("\n" + "="*80)
print("KALİTE DEĞERLENDİRMESİ")
print("="*80)

print(f"\n🎯 Pipeline Kalite Skoru:")

# Scoring sistemi
score = 0
max_score = 100

# 1. SGK Coverage (40 puan)
if sgk_coverage_pct >= 50:
    coverage_score = 40
    coverage_grade = "MÜKEMMEL"
elif sgk_coverage_pct >= 40:
    coverage_score = 30
    coverage_grade = "ÇOK İYİ"
elif sgk_coverage_pct >= 30:
    coverage_score = 20
    coverage_grade = "İYİ"
elif sgk_coverage_pct >= 20:
    coverage_score = 10
    coverage_grade = "ORTA"
else:
    coverage_score = 5
    coverage_grade = "DÜŞÜK"

score += coverage_score

# 2. Ürün Eşleşme Oranı (30 puan)
if product_match_pct >= 40:
    product_score = 30
elif product_match_pct >= 30:
    product_score = 20
elif product_match_pct >= 20:
    product_score = 10
else:
    product_score = 5

score += product_score

# 3. Tahmini Etkileşim Sayısı (30 puan)
if estimated_interactions >= 100000:
    interaction_score = 30
elif estimated_interactions >= 50000:
    interaction_score = 20
elif estimated_interactions >= 10000:
    interaction_score = 10
else:
    interaction_score = 5

score += interaction_score

print(f"\n  📊 Skor Dağılımı:")
print(f"    SGK Coverage ({sgk_coverage_pct:.1f}%): {coverage_score}/40 - {coverage_grade}")
print(f"    Ürün Eşleşme ({product_match_pct:.1f}%): {product_score}/30")
print(f"    Etkileşim Sayısı ({estimated_interactions:,}): {interaction_score}/30")
print(f"\n  🏆 TOPLAM SKOR: {score}/100")

# Genel değerlendirme
if score >= 80:
    verdict = "MÜKEMMEL - Pipeline yüksek kalitede sonuç üretecek!"
elif score >= 60:
    verdict = "İYİ - Pipeline kullanılabilir sonuçlar üretecek."
elif score >= 40:
    verdict = "ORTA - Pipeline çalışır ama iyileştirme gerekli."
else:
    verdict = "DÜŞÜK - Ciddi iyileştirme gerekli."

print(f"\n  ✅ {verdict}")

# ============================================================================
# AŞAMA 3.7: SGK-Only Ingredient'leri Göster (Manuel Mapping için)
# ============================================================================
print("\n" + "="*80)
print("SGK-ONLY INGREDIENT'LER (DDI'da Bulunmayan)")
print("="*80)

print(f"\n📝 Bu ingredient'ler için manuel Türkçe-İngilizce mapping gerekebilir...")
print(f"   Toplam SGK-only: {len(sgk_only):,}")

if len(sgk_only) > 0:
    # Frekansa göre sırala (en çok kullanılan ingredient'ler önce)
    sgk_only_freq = sgk_expanded_df[
        sgk_expanded_df['ingredient_en_norm'].isin(sgk_only)
    ]['ingredient_en_norm'].value_counts()
    
    print(f"\n  🔝 En Sık Kullanılan 20 SGK-Only Ingredient:")
    print(f"  (Bu ingredient'ler için manuel mapping en çok değer üretir)")
    print()
    for i, (ingredient, count) in enumerate(sgk_only_freq.head(20).items(), 1):
        print(f"    {i:>2}. {ingredient:<40} ({count:>3} ürün)")
    
    if len(sgk_only) > 20:
        print(f"\n    ... ve {len(sgk_only) - 20:,} tane daha")

# ============================================================================
# AŞAMA 3.8: Örnekler
# ============================================================================
print("\n" + "="*80)
print("ÖRNEK EŞLEŞMELER")
print("="*80)

print(f"\n📝 Örnek eşleşen ürünler (DDI'da etkileşim bulunacak):")
if len(sgk_matched) > 0:
    sample_matched = sgk_matched.head(5)
    for idx, row in sample_matched.iterrows():
        drug_name = str(row['drug_name'])[:50] if pd.notna(row['drug_name']) else 'N/A'
        ingredient = str(row['ingredient_en_norm']) if pd.notna(row['ingredient_en_norm']) else 'N/A'
        print(f"  ✓ {drug_name}")
        print(f"    → Ingredient: {ingredient}")

print(f"\n📝 Örnek eşleşmeyen ürünler (DDI'da yok):")
if len(sgk_unmatched) > 0:
    # Sadece ingredient'i olan ama match etmeyenleri göster
    unmatched_with_ingredient = sgk_unmatched[sgk_unmatched['ingredient_en_norm'].notna()]
    if len(unmatched_with_ingredient) > 0:
        sample_unmatched = unmatched_with_ingredient.head(5)
        for idx, row in sample_unmatched.iterrows():
            drug_name = str(row['drug_name'])[:50] if pd.notna(row['drug_name']) else 'N/A'
            ingredient = str(row['ingredient_en_norm']) if pd.notna(row['ingredient_en_norm']) else 'N/A'
            print(f"  ✗ {drug_name}")
            print(f"    → Ingredient: {ingredient} (DDI'da yok)")

# ============================================================================
# BLOK 3 ÖZET
# ============================================================================
print("\n" + "="*80)
print("✅ BLOK 3 TAMAMLANDI - KAPSAMA ANALİZİ BİTTİ")
print("="*80)

print(f"\n📊 Özet İstatistikler:")
print(f"\n  Ingredient Düzeyinde:")
print(f"    DDI Ingredient: {len(ddi_ingredients_set):,}")
print(f"    SGK Ingredient: {len(sgk_ingredients_set):,}")
print(f"    Örtüşen: {len(overlap):,} ({sgk_coverage_pct:.1f}%)")
print(f"\n  Ürün Düzeyinde:")
print(f"    Toplam SGK Ürün: {len(sgk_expanded_df):,}")
print(f"    Eşleşen Ürün: {product_match_count:,} ({product_match_pct:.1f}%)")
print(f"    Tahmini Etkileşim: {estimated_interactions:,}")
print(f"\n  Kalite Skoru: {score}/100 - {verdict}")

print(f"\n💡 Sonraki Blok: DDI Haritası Oluşturma ve Etkileşim Türetme")

# Karar noktası
if score < 40:
    print(f"\n⚠️  DİKKAT: Kalite skoru düşük!")
    print(f"   Devam etmeden önce missing data sorununu ele almalısınız.")
    print(f"   Veya düşük kapsama ile devam edip sonuçları görebilirsiniz.")
elif score < 60:
    print(f"\n⚠️  NOT: Orta kalite skoru.")
    print(f"   Pipeline çalışır ama manuel mapping ile iyileştirme önerilir.")
else:
    print(f"\n✅ Kalite skoru yeterli! Pipeline'a devam edebiliriz.")
````

````python
# ============================================================================
# BLOK 4: DDI HARİTASI OLUŞTURMA VE ETKİLEŞİM TÜRETME
# ============================================================================

print("\n" + "="*80)
print("BLOK 4: DDI HARİTASI VE ETKİLEŞİM TÜRETME")
print("="*80)

# ============================================================================
# AŞAMA 4.1: DDI Adjacency Map (Komşuluk Haritası) Oluştur
# ============================================================================
print("\n📊 DDI adjacency map oluşturuluyor...")
print("   Her etken madde için → etkileşimde bulunduğu diğer maddeler")

# Adjacency map: ingredient -> [(interacting_ingredient, interaction_type), ...]
ddi_map = defaultdict(list)

for _, row in ddi_df.iterrows():
    ing_a = row['ingredient_a_norm']
    ing_b = row['ingredient_b_norm']
    interaction_type = row['interaction_type']
    
    # Çift yönlü ekle (A→B ve B→A)
    # Çünkü eğer A, B ile etkileşirse, B de A ile etkileşir
    ddi_map[ing_a].append((ing_b, interaction_type))
    ddi_map[ing_b].append((ing_a, interaction_type))

print(f"✅ DDI haritası oluşturuldu")
print(f"  Etkileşimi olan ingredient sayısı: {len(ddi_map):,}")

# İstatistikler
interaction_counts = {ing: len(interactions) for ing, interactions in ddi_map.items()}
avg_interactions = sum(interaction_counts.values()) / len(interaction_counts)
max_interactions = max(interaction_counts.values())
min_interactions = min(interaction_counts.values())

print(f"\n📊 Etkileşim İstatistikleri:")
print(f"  Ortalama etkileşim/ingredient: {avg_interactions:.1f}")
print(f"  Maksimum etkileşim (tek ingredient): {max_interactions:,}")
print(f"  Minimum etkileşim: {min_interactions}")

# En çok etkileşimi olan ingredient'ler
print(f"\n🔝 En Çok Etkileşimi Olan 10 Ingredient:")
top_ingredients = sorted(interaction_counts.items(), key=lambda x: x[1], reverse=True)[:10]
for i, (ing, count) in enumerate(top_ingredients, 1):
    print(f"  {i:>2}. {ing:<40} {count:>5,} etkileşim")

# ============================================================================
# AŞAMA 4.2: Etkileşim Türetme (Inference) - ANA SÜREÇ!
# ============================================================================
print("\n" + "="*80)
print("ETKİLEŞİM TÜRETME (INFERENCE)")
print("="*80)

print(f"\n🚀 SGK ürünleri için etkileşimler türetiliyor...")
print(f"   Bu işlem birkaç saniye sürebilir...\n")

inferred_interactions = []
products_processed = 0
products_with_interactions = 0
products_without_interactions = 0
total_interactions = 0

# İlerleme takibi için
checkpoint = len(sgk_expanded_df) // 10  # Her %10'da rapor

for idx, row in sgk_expanded_df.iterrows():
    products_processed += 1
    
    # İlerleme göstergesi
    if checkpoint > 0 and products_processed % checkpoint == 0:
        progress = (products_processed / len(sgk_expanded_df)) * 100
        print(f"  İlerleme: {progress:.0f}% ({products_processed:,} / {len(sgk_expanded_df):,} ürün)")
    
    # Ürün bilgilerini al
    drug_name = row.get('drug_name', 'Unknown')
    ingredient = row['ingredient_en_norm']
    
    # Etken madde yoksa atla
    if pd.isna(ingredient):
        products_without_interactions += 1
        continue
    
    # Bu ingredient'in DDI haritasında etkileşimi var mı?
    if ingredient in ddi_map:
        products_with_interactions += 1
        
        # Bu ingredient için tüm etkileşimleri al
        interactions = ddi_map[ingredient]
        
        for interacting_ingredient, interaction_type in interactions:
            total_interactions += 1
            
            # Türetilmiş etkileşim kaydı oluştur
            inferred_interactions.append({
                'sgk_drug_name': drug_name,
                'ingredient_en_norm': ingredient,
                'interacts_with_ingredient_en_norm': interacting_ingredient,
                'interaction_type': interaction_type,
                'origin': 'inferred_from_global_ddi',
                'confidence': 'medium',
                'notes': 'Türetilmiş etkileşim - klinik olarak doğrulanmamıştır'
            })
    else:
        products_without_interactions += 1

# DataFrame'e çevir
tr_inferred_ddi_df = pd.DataFrame(inferred_interactions)

print(f"\n✅ Etkileşim türetme tamamlandı!")

# ============================================================================
# AŞAMA 4.3: Türetme Sonuçları
# ============================================================================
print("\n" + "="*80)
print("TÜRETME SONUÇLARI")
print("="*80)

print(f"\n📊 Süreç İstatistikleri:")
print(f"  Toplam işlenen ürün:          {products_processed:>10,}")
print(f"  Etkileşimi olan ürün:         {products_with_interactions:>10,}")
print(f"  Etkileşimi olmayan ürün:      {products_without_interactions:>10,}")
print(f"  Toplam türetilen etkileşim:   {total_interactions:>10,}")

if products_with_interactions > 0:
    avg_per_product = total_interactions / products_with_interactions
    print(f"  Ürün başına ort. etkileşim:   {avg_per_product:>10,.1f}")

# Gerçek vs Tahmin karşılaştırması
print(f"\n📊 Tahmin Karşılaştırması:")
print(f"  Tahmini etkileşim:            {estimated_interactions:>10,}")
print(f"  Gerçek etkileşim:             {total_interactions:>10,}")
difference_pct = ((total_interactions - estimated_interactions) / estimated_interactions * 100) if estimated_interactions > 0 else 0
print(f"  Fark:                         {total_interactions - estimated_interactions:>10,} ({difference_pct:+.1f}%)")

# ============================================================================
# AŞAMA 4.4: Türetilen Veri İncelemesi
# ============================================================================
print("\n" + "="*80)
print("TÜRETİLEN VERİ İNCELEMESİ")
print("="*80)

print(f"\n📊 Türetilen DataFrame Bilgileri:")
print(f"  Toplam satır:                 {len(tr_inferred_ddi_df):,}")
print(f"  Sütun sayısı:                 {len(tr_inferred_ddi_df.columns)}")
print(f"  Benzersiz SGK ürün:           {tr_inferred_ddi_df['sgk_drug_name'].nunique():,}")
print(f"  Benzersiz kaynak ingredient:  {tr_inferred_ddi_df['ingredient_en_norm'].nunique():,}")
print(f"  Benzersiz hedef ingredient:   {tr_inferred_ddi_df['interacts_with_ingredient_en_norm'].nunique():,}")
print(f"  Benzersiz etkileşim tipi:     {tr_inferred_ddi_df['interaction_type'].nunique():,}")

# En sık etkileşim tipleri
print(f"\n🔝 En Sık 10 Etkileşim Tipi:")
top_interaction_types = tr_inferred_ddi_df['interaction_type'].value_counts().head(10)
for i, (itype, count) in enumerate(top_interaction_types.items(), 1):
    pct = (count / len(tr_inferred_ddi_df)) * 100
    print(f"  {i:>2}. {itype[:50]:<50} {count:>7,} ({pct:>5.1f}%)")

# En çok etkileşimi olan ürünler
print(f"\n🔝 En Çok Etkileşimi Olan 10 Ürün:")
top_products = tr_inferred_ddi_df['sgk_drug_name'].value_counts().head(10)
for i, (product, count) in enumerate(top_products.items(), 1):
    print(f"  {i:>2}. {product[:60]:<60} {count:>4,} etkileşim")

# ============================================================================
# AŞAMA 4.5: Örnek Kayıtlar
# ============================================================================
print("\n" + "="*80)
print("ÖRNEK TÜRETİLMİŞ ETKİLEŞİMLER")
print("="*80)

print(f"\n📝 İlk 5 türetilmiş etkileşim:")
sample_records = tr_inferred_ddi_df.head(5)
for idx, row in sample_records.iterrows():
    print(f"\n  [{idx+1}]")
    print(f"  Ürün: {row['sgk_drug_name'][:60]}")
    print(f"  Etken Madde: {row['ingredient_en_norm']}")
    print(f"  Etkileşir: {row['interacts_with_ingredient_en_norm']}")
    print(f"  Tip: {row['interaction_type'][:80]}")

# Spesifik bir ürün için tüm etkileşimler (örnek)
if len(tr_inferred_ddi_df) > 0:
    sample_product = tr_inferred_ddi_df.iloc[0]['sgk_drug_name']
    product_interactions = tr_inferred_ddi_df[
        tr_inferred_ddi_df['sgk_drug_name'] == sample_product
    ]
    
    print(f"\n📝 Örnek Ürün için Tüm Etkileşimler:")
    print(f"   Ürün: {sample_product[:60]}")
    print(f"   Toplam etkileşim: {len(product_interactions):,}")
    print(f"\n   İlk 5 etkileşim:")
    for idx, row in product_interactions.head(5).iterrows():
        print(f"     • {row['interacts_with_ingredient_en_norm']}: {row['interaction_type'][:60]}")

# ============================================================================
# BLOK 4 ÖZET
# ============================================================================
print("\n" + "="*80)
print("✅ BLOK 4 TAMAMLANDI - ETKİLEŞİMLER TÜRETİLDİ!")
print("="*80)

print(f"\n🎉 BAŞARILARI:")
print(f"  ✓ DDI haritası oluşturuldu ({len(ddi_map):,} ingredient)")
print(f"  ✓ {products_with_interactions:,} ürün için etkileşim türetildi")
print(f"  ✓ {total_interactions:,} etkileşim kaydı oluşturuldu")
print(f"  ✓ tr_inferred_ddi DataFrame hazır")

print(f"\n📊 Ana Deliverable:")
print(f"  DataFrame adı: tr_inferred_ddi_df")
print(f"  Boyut: {len(tr_inferred_ddi_df):,} satır × {len(tr_inferred_ddi_df.columns)} sütun")
print(f"  Memory: ~{tr_inferred_ddi_df.memory_usage(deep=True).sum() / 1024 / 1024:.1f} MB")

print(f"\n💡 Sonraki Blok: Kalite Kontrol ve Temizlik")
````

````python
# ============================================================================
# BLOK 4.5: DETAYLI VERİ ANALİZİ VE DOĞRULAMA
# ============================================================================

print("\n" + "="*80)
print("BLOK 4.5: DETAYLI VERİ ANALİZİ VE DOĞRULAMA")
print("="*80)

# ============================================================================
# AŞAMA 4.5.1: Mevcut Veriyi İncele
# ============================================================================
print("\n📊 TÜRETİLEN VERİNİN DETAYLI İNCELEMESİ")
print("="*80)

print(f"\n1️⃣ TEMEL BİLGİLER:")
print(f"   DataFrame adı: tr_inferred_ddi_df")
print(f"   Toplam satır: {len(tr_inferred_ddi_df):,}")
print(f"   Toplam sütun: {len(tr_inferred_ddi_df.columns)}")
print(f"   Memory kullanımı: {tr_inferred_ddi_df.memory_usage(deep=True).sum() / 1024 / 1024:.2f} MB")

print(f"\n2️⃣ SÜTUN YAPISI:")
print(f"   Sütunlar ve tipleri:")
for col in tr_inferred_ddi_df.columns:
    dtype = tr_inferred_ddi_df[col].dtype
    null_count = tr_inferred_ddi_df[col].isnull().sum()
    null_pct = (null_count / len(tr_inferred_ddi_df)) * 100
    print(f"     • {col:<45} {str(dtype):<15} (Null: {null_count:>7,} - {null_pct:>5.2f}%)")

print(f"\n3️⃣ BENZERSIZ DEĞER ANALİZİ:")
print(f"   • Benzersiz SGK ürün:               {tr_inferred_ddi_df['sgk_drug_name'].nunique():>10,}")
print(f"   • Benzersiz kaynak ingredient:      {tr_inferred_ddi_df['ingredient_en_norm'].nunique():>10,}")
print(f"   • Benzersiz hedef ingredient:       {tr_inferred_ddi_df['interacts_with_ingredient_en_norm'].nunique():>10,}")
print(f"   • Benzersiz etkileşim tipi:         {tr_inferred_ddi_df['interaction_type'].nunique():>10,}")

print(f"\n4️⃣ ÖRNEK SATIRLAR (İlk 3):")
print(tr_inferred_ddi_df.head(3))

# ============================================================================
# AŞAMA 4.5.2: Veri Kalitesi Kontrolleri
# ============================================================================
print("\n" + "="*80)
print("VERİ KALİTESİ KONTROL")
print("="*80)

# Kontrol 1: Null değerler var mı?
print(f"\n🔍 KONTROL 1: NULL DEĞER ANALİZİ")
null_summary = tr_inferred_ddi_df.isnull().sum()
if null_summary.sum() == 0:
    print(f"   ✅ Hiç null değer yok - Mükemmel!")
else:
    print(f"   ⚠️ Null değerler bulundu:")
    for col, count in null_summary[null_summary > 0].items():
        print(f"     • {col}: {count:,} null değer")

# Kontrol 2: Self-interaction var mı? (ingredient kendisiyle etkileşiyor mu?)
print(f"\n🔍 KONTROL 2: SELF-INTERACTION KONTROLÜ")
self_interactions = tr_inferred_ddi_df[
    tr_inferred_ddi_df['ingredient_en_norm'] == 
    tr_inferred_ddi_df['interacts_with_ingredient_en_norm']
]
if len(self_interactions) == 0:
    print(f"   ✅ Self-interaction yok - Doğru!")
else:
    print(f"   ⚠️ {len(self_interactions):,} adet self-interaction bulundu!")
    print(f"   Örnek:")
    print(self_interactions[['ingredient_en_norm', 'interacts_with_ingredient_en_norm']].head(3))

# Kontrol 3: Duplicate kayıt var mı?
print(f"\n🔍 KONTROL 3: DUPLICATE KAYIT KONTROLÜ")
duplicates = tr_inferred_ddi_df.duplicated(subset=[
    'sgk_drug_name', 
    'ingredient_en_norm', 
    'interacts_with_ingredient_en_norm', 
    'interaction_type'
]).sum()
if duplicates == 0:
    print(f"   ✅ Duplicate kayıt yok - Temiz!")
else:
    print(f"   ⚠️ {duplicates:,} adet duplicate kayıt var!")

# Kontrol 4: Etkileşim tipi dağılımı mantıklı mı?
print(f"\n🔍 KONTROL 4: ETKİLEŞİM TİPİ DAĞILIMI")
interaction_type_dist = tr_inferred_ddi_df['interaction_type'].value_counts()
print(f"   Toplam benzersiz tip: {len(interaction_type_dist)}")
print(f"   En sık 5 tip:")
for itype, count in interaction_type_dist.head(5).items():
    pct = (count / len(tr_inferred_ddi_df)) * 100
    print(f"     • {itype[:60]:<60} {count:>7,} ({pct:>5.1f}%)")

# Kontrol 5: Ürün başına etkileşim dağılımı
print(f"\n🔍 KONTROL 5: ÜRÜN BAŞINA ETKİLEŞİM DAĞILIMI")
interactions_per_product = tr_inferred_ddi_df['sgk_drug_name'].value_counts()
print(f"   Ortalama etkileşim/ürün: {interactions_per_product.mean():.1f}")
print(f"   Medyan etkileşim/ürün: {interactions_per_product.median():.0f}")
print(f"   Min etkileşim: {interactions_per_product.min()}")
print(f"   Max etkileşim: {interactions_per_product.max():,}")
print(f"\n   Dağılım:")
print(f"     • 1-50 etkileşim:    {(interactions_per_product <= 50).sum():>6,} ürün")
print(f"     • 51-100 etkileşim:  {((interactions_per_product > 50) & (interactions_per_product <= 100)).sum():>6,} ürün")
print(f"     • 101-200 etkileşim: {((interactions_per_product > 100) & (interactions_per_product <= 200)).sum():>6,} ürün")
print(f"     • 201+ etkileşim:    {(interactions_per_product > 200).sum():>6,} ürün")

# ============================================================================
# AŞAMA 4.5.3: Mantıksal Tutarlılık Kontrolleri
# ============================================================================
print("\n" + "="*80)
print("MANTIKSAL TUTARLILIK KONTROL")
print("="*80)

# Spesifik bir ürün seç ve detaylı incele
print(f"\n🔬 ÖRNEK ÜRÜN DETAYLI İNCELEME:")
if len(tr_inferred_ddi_df) > 0:
    # En çok etkileşimi olan ürünü seç
    sample_product = interactions_per_product.index[0]
    product_data = tr_inferred_ddi_df[tr_inferred_ddi_df['sgk_drug_name'] == sample_product]
    
    print(f"\n   Seçilen Ürün: {sample_product[:70]}")
    print(f"   Etken Madde: {product_data.iloc[0]['ingredient_en_norm']}")
    print(f"   Toplam Etkileşim: {len(product_data):,}")
    print(f"\n   İlk 10 Etkileşim:")
    
    for idx, row in product_data.head(10).iterrows():
        print(f"     {idx+1}. {row['interacts_with_ingredient_en_norm']:<30}")
        print(f"        → {row['interaction_type'][:70]}")

# Origin ve Confidence kontrol
print(f"\n🔬 METADATA KONTROL:")
print(f"   Origin değerleri:")
if 'origin' in tr_inferred_ddi_df.columns:
    origin_counts = tr_inferred_ddi_df['origin'].value_counts()
    for origin, count in origin_counts.items():
        print(f"     • {origin}: {count:,} kayıt")
    
    # Hepsi 'inferred_from_global_ddi' olmalı
    if len(origin_counts) == 1 and 'inferred_from_global_ddi' in origin_counts.index:
        print(f"   ✅ Tüm kayıtlar doğru origin'e sahip")

print(f"\n   Confidence değerleri:")
if 'confidence' in tr_inferred_ddi_df.columns:
    confidence_counts = tr_inferred_ddi_df['confidence'].value_counts()
    for conf, count in confidence_counts.items():
        print(f"     • {conf}: {count:,} kayıt")
    
    # Hepsi 'medium' olmalı
    if len(confidence_counts) == 1 and 'medium' in confidence_counts.index:
        print(f"   ✅ Tüm kayıtlar doğru confidence'a sahip")

# ============================================================================
# AŞAMA 4.5.4: Veriyi Kaydet ve Geri Oku
# ============================================================================
print("\n" + "="*80)
print("VERİYİ KAYDETME VE GERİ OKUMA")
print("="*80)

# CSV olarak kaydet
output_filename = 'tr_inferred_ddi.csv'
print(f"\n💾 Veri kaydediliyor: {output_filename}")

tr_inferred_ddi_df.to_csv(output_filename, index=False, encoding='utf-8')
print(f"   ✅ Kayıt başarılı!")

# Dosya boyutunu kontrol et
import os
file_size_mb = os.path.getsize(output_filename) / 1024 / 1024
print(f"   📊 Dosya boyutu: {file_size_mb:.2f} MB")

# Geri oku ve doğrula
print(f"\n📖 Veri geri okunuyor...")
tr_inferred_ddi_loaded = pd.read_csv(output_filename, encoding='utf-8')
print(f"   ✅ Okuma başarılı!")

print(f"\n🔍 KAYIT-OKUMA DOĞRULAMA:")
print(f"   Orijinal satır sayısı: {len(tr_inferred_ddi_df):,}")
print(f"   Okunan satır sayısı:   {len(tr_inferred_ddi_loaded):,}")
print(f"   Orijinal sütun sayısı: {len(tr_inferred_ddi_df.columns)}")
print(f"   Okunan sütun sayısı:   {len(tr_inferred_ddi_loaded.columns)}")

if len(tr_inferred_ddi_df) == len(tr_inferred_ddi_loaded):
    print(f"   ✅ Satır sayıları eşleşiyor!")
else:
    print(f"   ⚠️ Satır sayıları farklı!")

if list(tr_inferred_ddi_df.columns) == list(tr_inferred_ddi_loaded.columns):
    print(f"   ✅ Sütunlar eşleşiyor!")
else:
    print(f"   ⚠️ Sütunlar farklı!")

# Örnek kayıtları karşılaştır
print(f"\n📋 İLK 3 SATIR (Kaydedilen Dosyadan):")
print(tr_inferred_ddi_loaded.head(3))

# ============================================================================
# AŞAMA 4.5.5: İstatistiksel Özet
# ============================================================================
print("\n" + "="*80)
print("İSTATİSTİKSEL ÖZET")
print("="*80)

print(f"\n📊 VERİ SETİ KAPSAMLI ÖZETİ:")
print(f"\n1. BOYUT BİLGİLERİ:")
print(f"   • Toplam etkileşim kaydı:        {len(tr_inferred_ddi_df):>12,}")
print(f"   • Benzersiz ürün sayısı:         {tr_inferred_ddi_df['sgk_drug_name'].nunique():>12,}")
print(f"   • Benzersiz kaynak ingredient:   {tr_inferred_ddi_df['ingredient_en_norm'].nunique():>12,}")
print(f"   • Benzersiz hedef ingredient:    {tr_inferred_ddi_df['interacts_with_ingredient_en_norm'].nunique():>12,}")
print(f"   • Benzersiz etkileşim tipi:      {tr_inferred_ddi_df['interaction_type'].nunique():>12,}")

print(f"\n2. KAPSAMA BİLGİLERİ:")
original_sgk_count = len(sgk_df)
matched_product_count = tr_inferred_ddi_df['sgk_drug_name'].nunique()
coverage_pct = (matched_product_count / original_sgk_count) * 100
print(f"   • Orijinal SGK ürün sayısı:      {original_sgk_count:>12,}")
print(f"   • Etkileşimi olan ürün:          {matched_product_count:>12,}")
print(f"   • Kapsama oranı:                 {coverage_pct:>11.2f}%")

print(f"\n3. ORTALAMA İSTATİSTİKLER:")
avg_interactions = len(tr_inferred_ddi_df) / matched_product_count if matched_product_count > 0 else 0
print(f"   • Ürün başına ort. etkileşim:    {avg_interactions:>12.1f}")
print(f"   • Karakter/satır (ortalama):     {tr_inferred_ddi_df.memory_usage(deep=True).sum() / len(tr_inferred_ddi_df):>12.0f} bytes")

print(f"\n4. KALİTE METRİKLERİ:")
null_total = tr_inferred_ddi_df.isnull().sum().sum()
duplicate_pct = (duplicates / len(tr_inferred_ddi_df) * 100) if len(tr_inferred_ddi_df) > 0 else 0
print(f"   • Null değer oranı:              {(null_total / (len(tr_inferred_ddi_df) * len(tr_inferred_ddi_df.columns)) * 100):>11.2f}%")
print(f"   • Duplicate oranı:               {duplicate_pct:>11.2f}%")
print(f"   • Veri bütünlüğü skoru:          {(100 - null_total / len(tr_inferred_ddi_df) * 100):>11.2f}%")

# ============================================================================
# BLOK 4.5 ÖZET
# ============================================================================
print("\n" + "="*80)
print("✅ BLOK 4.5 TAMAMLANDI - VERİ ANALİZİ VE KAYIT BİTTİ")
print("="*80)

print(f"\n🎉 ÖNEMLİ BAŞARILAR:")
print(f"   ✓ {len(tr_inferred_ddi_df):,} etkileşim kaydı üretildi")
print(f"   ✓ {matched_product_count:,} ürün için etkileşim türetildi")
print(f"   ✓ Veri kalite kontrolleri yapıldı")
print(f"   ✓ {output_filename} olarak kaydedildi ({file_size_mb:.2f} MB)")
print(f"   ✓ Veri geri okundu ve doğrulandı")

print(f"\n📊 VERİ SETİ ÖZELLİKLERİ:")
print(f"   • Format: CSV")
print(f"   • Encoding: UTF-8")
print(f"   • Boyut: {len(tr_inferred_ddi_df):,} satır × {len(tr_inferred_ddi_df.columns)} sütun")
print(f"   • Disk Boyutu: {file_size_mb:.2f} MB")
print(f"   • Memory: {tr_inferred_ddi_df.memory_usage(deep=True).sum() / 1024 / 1024:.2f} MB")

print(f"\n⚠️ ÖNEMLİ NOTLAR:")
print(f"   • Bu veriler TÜRETİLMİŞTİR (inferred)")
print(f"   • Klinik olarak DOĞRULANMAMIŞTIR")
print(f"   • Confidence seviyesi: MEDIUM")
print(f"   • Referans amaçlı kullanılmalıdır")

print(f"\n💡 Sonraki Blok: Kalite Kontrol ve Final Temizlik")
````
