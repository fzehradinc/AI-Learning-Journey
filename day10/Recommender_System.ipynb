```python
mg = (rating.groupby('title')['rating']
      .agg(size='count', sum='sum', mean='mean')
      .reset_index())

pm = mg.sort_values('mean', ascending=False)

gs = pm['sum'].sum()     # ✅ toplam rating puanı
gc = pm['size'].sum()    # ✅ toplam oy sayısı
gmean = gs / gc          # ✅ genel ortalama

print("Toplam puan (sum):", gs)
print("Toplam oy (count):", gc)
print("Genel ortalama:", gmean)
```

```python
import pandas as pd
import numpy as np

# rating DataFrame'in hazır olduğunu varsayıyorum:
# rating kolonları: ['user_id','item_id','rating','timestamp','title'] (veya en azından 'title' ve 'rating' olmalı)

# 1) Film bazında istatistikler (popülerlik + kalite)
mg = (rating.groupby('title')['rating']
      .agg(size='count', sum='sum', mean='mean')
      .reset_index())

# 2) Ortalama puana göre sırala (istersen önce filtrele)
pm = mg.sort_values('mean', ascending=False)

# 3) Toplam puan (tüm filmlerin sum toplamı)
gs = pm['sum'].sum()

# 4) Yüzde katkı: her filmin toplam puan içindeki payı (%)
pm['percentage'] = (pm['sum'] / gs) * 100

# 5) Rank: en yüksek yüzde = en iyi rank (1 en iyi)
pm['rank'] = pm['percentage'].rank(ascending=False, method='dense')

# 6) En popüler (rank küçükten büyüğe)
top_popular = pm.sort_values('rank', ascending=True)

# Göster
top_popular.head(20)
```
