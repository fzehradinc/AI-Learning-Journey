```python



import pandas as pd

ratings = pd.read_csv(
    "ml-100k/u.data",
    sep="\t",
    names=["user_id", "item_id", "rating", "timestamp"]
)

ratings.head()
```

```python
movies=pd.read_csv('ml-100k/u.item', encoding='iso-8859-1', sep='|', header=None, names=['movie_id', 'title'], usecols=range(2))
```

```python
movies
```

```python
user=pd.read_csv('ml-100k/u.user', sep='|', header=None, names=['user_id','Age','Gender','Profession', 'Zipcode'])
```

```python
user
```

```python
rating=pd.merge(movies,ratings)
```

```python
print("ratings:", ratings.shape, ratings.columns)
print("movies :", movies.shape, movies.columns)

display(ratings.head())
display(movies.head())
```

```python
rating = rating.drop(columns=["movie_id"])
```

```python
rating = pd.merge(
    ratings,
    movies,
    left_on="item_id",
    right_on="movie_id",
    how="inner"
)

rating.head()
```

```python
rating=pd.merge(rating,user)
```

```python
rating
```

```python
print("ratings:", ratings.shape, ratings.columns)
print("user :", user.shape, user.columns)

display(ratings.head())
display(user.head())
```

```python
mf=rating.pivot_table(index=['user_id'], columns=['title'], values='rating')
```

```python
mf.head()
```

```python
swr=mf['Star Wars (1977)']
```

```python
swr.value_counts()
```

```python
mf[['Star Wars (1977)', '101 Dalmatians (1996)']].corr()
```

```python
sm=mf.corrwith(swr)
```

```python
sm.sort_values(ascending=False)
```

```python
import numpy as np
```

```python
mstats=rating.groupby('title').agg({'rating':[np.size, np.mean]})
```

```python
mstats.head()
```

```python
pm=mstats[mstats['rating']['size']>100]
```

```python
smdata=pd.DataFrame(sm, columns=['similarity'])
```

```python
smdata
```

```python
pm.columns=pm.columns.get_level_values(0)
df=pm.join(smdata)
```

```python
df.sort_values(['similarity'], ascending=False
```
