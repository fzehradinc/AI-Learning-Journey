```python
import pandas as pd
```

```python
data={'Age': [25,30,35,40,None],
      'Gender': ['Male','Female','Male','Female','Female'],  
      'Salary':[5000,60000,70000,None,90000]
     }
```

```python
df=pd.DataFrame(data)
```

```python
df['Age']=df['Age'].fillna(df.Age.mean())
```

```python
df['Salary']=df['Salary'].fillna(df.Salary.mode()[0])
```

```python
x=df[['Age','Gender']]
y=df[['Salary']]
```

```python
x=pd.get_dummies(x,drop_first=True)
```

```python
x
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
x=sc.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y, random_state=42, test_size=.20)
```

```python
import sklearn, sys
print("Python:", sys.executable)
print("sklearn version:", sklearn.__version__)
print("sklearn path:", sklearn.__file__)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt=DecisionTreeClassifier()
```

```python
model=dt.fit(x_train,y_train)
```

```python
tahmin=model.predict(x)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y,tahmin)
```

```python
yd=[[43, False]]
```

```python
model.predict(yd)
```

```python
data={'Age':[25,30,35,40,53,32,45,22,47,55],
      'Gender':['Male','Female','Male','Female','Female','Male','Female','Male','Female','Female'],
      'Salary':[5000,60000,70000,23000,90000,50000,60000,72000,65000,90000],
      'Purchased':[0,0,1,1,1,1,1,0,0,1]
     }
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dc=DecisionTreeClassifier()
```

```python
x=df[['Age','Gender','Salary']]
y=df[['Purchased']]
```

```python
x=pd.get_dummies(x,drop_first=True)
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y, random_state=42, test_size=.20)
```

```python
model=dc.fit(x_train,y_train)
```

```python
tahmin=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test, tahmin)
```
