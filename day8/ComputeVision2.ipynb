```python

#where is the waldo - polis topluluk icerisinde aradigi kisiyi bulmak istiyor
image=cv2.imread('WaldoBeach.jpg')
cv2.imshow('where is waldo?', image)

gray=cv2.imread('WaldoBeach.jpg',0)
waldo=cv2.imread('Waldo.jpg',0)

result=cv2.matchTemplate(gray, waldo, cv2.TM_CCOEFF)
minval, maxval, minloc, maxloc=cv2.minMaxLoc(result)

top_left=maxloc
bottom_right=(top_left[0]+50,top_left[1]+50)
cv2.rectangle(image, top_left, bottom_right, (0,0,255),5)

cv2.imshow('where is waldo?', image)

cv2.waitKey()
cv2.destroyAllWindows()
```

```python
import cv2

# Kamerayı başlat
cap = cv2.VideoCapture(0)

# İlk iki frame'i oku
ret1, frame1 = cap.read()
ret2, frame2 = cap.read()

while True:
    # Frame'leri gri tona çevir
    frame1sb = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
    frame2sb = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

    # İki frame arasındaki farkı al (hareket tespiti)
    diff = cv2.absdiff(frame1sb, frame2sb)

    # Sonucu göster
    cv2.imshow("Motion", diff)

    # Frame'leri kaydır
    frame1 = frame2
    ret, frame2 = cap.read()

    if not ret:
        cap.release()
        break

    # Klavye kontrolü
    key = cv2.waitKey(10)
    if key == ord("q"):
        cap.release()
        break

# Tüm pencereleri kapat
cv2.destroyAllWindows()
```
