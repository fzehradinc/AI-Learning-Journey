```python

import cv2

# Kamerayı başlat
cap = cv2.VideoCapture(0)

# İlk iki frame'i oku
ret1, frame1 = cap.read()
ret2, frame2 = cap.read()

while True:
    # Frame'leri gri tona çevir
    frame1sb = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
    frame2sb = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

    # İki frame arasındaki farkı al (hareket tespiti)
    diff = cv2.absdiff(frame1sb, frame2sb)

    # Sonucu göster
    cv2.imshow("Motion", diff)

    # Frame'leri kaydır
    frame1 = frame2
    ret, frame2 = cap.read()

    if not ret:
        cap.release()
        break

    # Klavye kontrolü
    key = cv2.waitKey(10)
    if key == ord("q"):
        cap.release()
        break

# Tüm pencereleri kapat
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

cap = cv2.VideoCapture("airplanes.mp4")

ret1, frame1 = cap.read()
ret2, frame2 = cap.read()

while True:
    frame1_gray = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
    frame2_gray = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

    frame1_blur = cv2.GaussianBlur(frame1_gray, (21, 21), 0)
    frame2_blur = cv2.GaussianBlur(frame2_gray, (21, 21), 0)

    diff = cv2.absdiff(frame1_blur, frame2_blur)

    thresh = cv2.threshold(diff, 20, 255, cv2.THRESH_BINARY)[1]
    final = cv2.dilate(thresh, None, iterations=2)

    masked = cv2.bitwise_and(frame1, frame1, mask=thresh)
    white_pixels = np.sum(thresh) / 255

    rows, cols = thresh.shape
    total = rows * cols

    if white_pixels > 0.01 * total:
        font = cv2.FONT_HERSHEY_SIMPLEX
        cv2.putText(
            frame1,
            "Movement Detected - Hareket Var",
            (10, 50),
            font,
            1,
            (0, 0, 255),
            2,
            cv2.LINE_AA
        )

    cv2.imshow("Motion", frame1)

    frame1 = frame2
    ret, frame2 = cap.read()

    if not ret:
        break

    key = cv2.waitKey(10)
    if key == 27 or key == ord("q"):
        break

cv2.destroyAllWindows()
cap.release()
```

```python
import cv2

# Load the pre-trained face detection model
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

def face_extractor(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)
    
    # If no face detected, return the input image
    if len(faces) == 0:
        return None
    
    # Extract faces
    for (x, y, w, h) in faces:
        extracted_face = img[y:y+h, x:x+w]
        return extracted_face

# Initialize video capture
cap = cv2.VideoCapture(0)

count = 0

while True:
    # Read a frame from the capture
    ret, frame = cap.read()
    
    if frame is not None:
        face = face_extractor(frame)
        
        if face is not None:
            count += 1
            face = cv2.resize(face, (200, 200))
            
            # Save the extracted face as an image
            file_name = f"face_{count}.jpeg"
            cv2.imwrite(file_name, face)
            
            # Display the extracted face
            cv2.putText(face, f"Count: {count}", (10, 20), cv2.FONT_HERSHEY_COMPLEX, 0.5, (255, 255, 255), 1)
            cv2.imshow('Face Extractor', face)
        
    # Exit if 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release the capture and close windows
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

# Yüz algılama modelini yükle
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

def face_extractor(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)
    
    # Eğer yüz algılanmadıysa, girdi resmini döndür
    if len(faces) == 0:
        return None
    
    # Yüzü çıkar
    for (x, y, w, h) in faces:
        extracted_face = img[y:y+h, x:x+w]
        return extracted_face

# Video yakalamayı başlat
cap = cv2.VideoCapture(0)

count = 0
target_count = 300  # Hedef resim sayısı

while count < target_count:
    ret, frame = cap.read()
    
    if frame is not None:
        face = face_extractor(frame)
        
        if face is not None:
            count += 1
            # Yüzü yeniden boyutlandır ve kaydet
            face = cv2.resize(face, (200, 200))
            
            # Yüzü dosyaya kaydet
            file_name = f"face_{count}.jpeg"
            cv2.imwrite(file_name, face)
            
            # Çıkarılan yüzü göster
            cv2.putText(face, f"Count: {count}", (10, 20), cv2.FONT_HERSHEY_COMPLEX, 0.5, (255, 255, 255), 1)
            cv2.imshow('Face Extractor', face)
        
    # 300 resim kaydedildiğinde döngüden çık
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Yakalamayı serbest bırak ve pencereleri kapat
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import os

# Kaydetmek istediğiniz dizin
data_path = 'faces'  # Yüzlerin kaydedileceği dizin

# Dizin yoksa oluştur
if not os.path.exists(data_path):
    os.makedirs(data_path)

# Yüz algılama modelini yükle
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

def face_extractor(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)

    if len(faces) == 0:
        return None

    for (x, y, w, h) in faces:
        extracted_face = img[y:y+h, x:x+w]
        return extracted_face

# Video yakalamayı başlat
cap = cv2.VideoCapture(0)

count = 0
target_count = 300  # Hedef resim sayısı

while count < target_count:
    ret, frame = cap.read()

    if frame is not None:
        face = face_extractor(frame)

        if face is not None:
            count += 1
            face = cv2.resize(face, (200, 200))

            # Yüzü dosyaya kaydet
            file_name = os.path.join(data_path, f"face_{count}.jpeg")
            cv2.imwrite(file_name, face)

            # Çıkarılan yüzü göster
            cv2.putText(face, f"Count: {count}", (10, 20), cv2.FONT_HERSHEY_COMPLEX, 0.5, (255, 255, 255), 1)
            cv2.imshow('Face Extractor', face)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Yakalamayı serbest bırak ve pencereleri kapat
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import os

# Yüz verilerinin saklandığı dizin
data_path = 'faces'  # Yüzlerin kaydedildiği dizin
only_files = os.listdir(data_path)

# Eğitim verilerini ve etiketlerini saklamak için listeler
X = []
y = []

# Her bir yüz görüntüsünü yükleyin
for i, file_name in enumerate(only_files):
    image_path = os.path.join(data_path, file_name)
    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)  # Görüntüyü gri tonlamada yükle
    
    if image is not None:
        X.append(image)
        y.append(i)  # Etiket olarak dizinin sırasını kullan
        
# Verileri NumPy dizisine dönüştür
X = np.array(X)
y = np.array(y)

# Yüz tanıma modelini eğitme
model = cv2.face.LBPHFaceRecognizer_create()  # LBPH (Local Binary Patterns Histograms) tanıyıcı
model.train(X, y)  # Modeli eğit

# Modeli kaydet
model.save('face_recognition_model.xml')

print("Model başarıyla eğitildi ve kaydedildi.")
```

```python
import cv2
import numpy as np

# Eğitimli modelin yüklenmesi
model = cv2.face.LBPHFaceRecognizer_create()
model.read('face_recognition_model.xml')

# Yüz algılama modelini yükle
face_cascade = cv2.CascadeClassifier('haarcascade_frontalface_default.xml')

# Video yakalamayı başlat
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    
    if frame is not None:
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)

        for (x, y, w, h) in faces:
            roi_gray = gray[y:y+h, x:x+w]
            label, confidence = model.predict(roi_gray)

            # Güven seviyesini kontrol et
            if confidence < 100:  # Güven seviyesi threshold
                label_text = f'User {label}'
            else:
                label_text = 'Unknown'
            
            # Çerçeve çizme ve metin yazma
            cv2.rectangle(frame, (x, y), (x + w, y + h), (255, 0, 0), 2)
            cv2.putText(frame, label_text, (x, y-10), cv2.FONT_HERSHEY_COMPLEX, 0.8, (255, 255, 255), 2)

        cv2.imshow('Face Recognition', frame)

    # 'q' tuşuna basarak çıkış yapma
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Video yakalamayı serbest bırak ve pencereleri kapat
cap.release()
cv2.destroyAllWindows()
```
